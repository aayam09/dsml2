### Descriptive Statistics of Age and Fare

The summary statistics provide an overview of the **Age** and **Fare** variables in the Titanic dataset.

| Statistic | Age | Fare |
|----------|---------:|---------:|
| Count | 714 | 891 |
| Mean | 29.70 | 32.20 |
| Standard Deviation | 14.53 | 49.69 |
| Minimum | 0.42 | 0.00 |
| 25th Percentile | 20.13 | 7.91 |
| Median (50%) | 28.00 | 14.45 |
| 75th Percentile | 38.00 | 31.00 |
| Maximum | 80.00 | 512.33 |

**Observation:**

- The **Age** column contains **714** non-null values, indicating that some passenger ages are missing, whereas the **Fare** column has no missing values.
- The average passenger age is **29.7 years**, with most passengers being between **20 and 38 years** old.
- Passenger fares vary considerably, with an average fare of **32.20** and values ranging from **0.00** to **512.33**.
- The high standard deviation of the **Fare** column suggests a wide variation in ticket prices, likely due to differences in passenger class and ticket type.

### Distribution of Passengers by Sex and Survival Status

This analysis presents the proportion of passengers based on their gender and survival status.

| Sex | Survived | Proportion |
|------|----------|-----------:|
| Male | No (0) | 52.53% |
| Female | Yes (1) | 26.15% |
| Male | Yes (1) | 12.23% |
| Female | No (0) | 9.09% |

**Observation:**

- The largest group in the dataset consists of **male passengers who did not survive (52.53%)**.
- **Female survivors account for 26.15%** of the total passengers, which is higher than the proportion of male survivors (12.23%).
- Overall, female passengers exhibited a higher survival rate than male passengers, indicating that gender was a significant factor influencing survival.
