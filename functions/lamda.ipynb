## Python Lambda Expression

```python
def add(x,y):
    return x+y

add(4,5)
```

```python
add_lamda = lambda x,y:x+y
```

```python
type(add_lamda)
```

```python
add_lamda(4,5)
```

```python
Area = lambda a,b,c:2 * (a*b + b*c + c*a)
Area(4,4,4)
```

### Map Function

```python
discount = lambda x:x*0.9 if x>100 else x
prices = [45, 120, 200, 75, 150]
dis_list = map(discount,prices)
list(dis_list)
```

```python
large_obj = range(1,10000000)
res_value = []
```
