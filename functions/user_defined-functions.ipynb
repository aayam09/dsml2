## User defined functions

```python
def sum_of_list(inp_list):
    result = 0

    for x in inp_list:
        result += x
        
    return result
```

```python
sum_of_list([2,5,8,5])
```

```python
def greet(name, greetings="Morning"):
    return f"{greetings}, {name}"
```

```python
greet("Rohit")
```

```python
greet("Bob","Hi")
```

```python
def print_args(*args):
    print("Positional Args")
    for arg in args:
        print(arg)
```

```python
print_args(1,34,3,[4,8],5,"aayam")
```

```python
def print_kargs(**kwargs):
    print("keyword Args")
    print(kwargs)
```

```python
print_kargs(length=5, breadth=5)
```

```python
def print_arguements(*args,**kwargs):
    print("Positional")
    print(args)
    print("-" * 50)
    print(kwargs)
```

```python
print_arguements("ram",1,True, score =[1,2,3], greet = "Hello")
```

### Documentation in Fuction

```python
def square(n: int) -> int:
    """
    Calculate the square of a number.

    Parameters
    ----------
    n : int
        The integer value to be squared.

    Returns
    -------
    int
        The square of the input integer.

    Examples
    --------
    >>> square(4)
    16

    >>> square(-3)
    9
    """
    return n ** 2
```

```python
square(3)
```

```python
square()
```
