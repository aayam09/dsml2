## User defined functions

```python
def sum_of_list(inp_list):
    result = 0

    for x in inp_list:
        result += x
        
    return result
```

```python
sum_of_list([2,5,8,5])
```

```python
def print_args(*args):
    print("Positional Args")
    for arg in args:
        print(arg)
```

```python
def square(n: int) -> int:
    """
    Calculate the square of a number.

    Parameters
    ----------
    n : int
        The integer value to be squared.

    Returns
    -------
    int
        The square of the input integer.
    """
    return n ** 2
```

```python
import math
import statistics

number_list = (2, 3, 4, 5, 6)

def standard_deviation(*number) -> float:
    """Calculate the standard deviation of a list of numbers.
    Args:
        number_list (list): A list of numbers.
        Returns:
        float: The standard deviation of the numbers in the list.
    Raises:
        TypeError: If the input is not a list.
            """
    if not isinstance(number_list, list):
        raise TypeError("Input must be a list")      

    mean = statistics.mean(number_list)
    total = 0
    for x in number_list:
        total += (x - mean) ** 2
    return math.sqrt(total / len(number_list))  # population standard deviation

print(standard_deviation(number_list))
```

```python
import math
import statistics

def standard_deviation(*numbers) -> float:
    """
    Calculate the population standard deviation of multiple numbers.

    Parameters
    ----------
    *numbers : float
        Variable number of numeric arguments.

    Returns
    -------
    float
        The population standard deviation.
    """
    mean = statistics.mean(numbers)

    total = 0
    for x in numbers:
        total += (x - mean) ** 2

    return math.sqrt(total / len(numbers))


print(standard_deviation(2, 3, 4, 5, 6,5,3,767,5567,33))
```
