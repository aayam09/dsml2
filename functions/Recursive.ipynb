## Recursive Functions

```python
def fib(n):
    if n == 0:
        return 0
    elif n==1:
        return 1
    else:
        return fib(n-1)+ fib(n-2)
```

```python
%%time
fib(30)
```

```python
cache = {0 :0,1: 1}

def fib(n):
    if n in cache:
        return cache[n]
    else:
        result = fib(n-1) + fib(n-2)
        cache[n] = result
        return result
```

```python
%%time
fib(500)
```
