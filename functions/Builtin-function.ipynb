### open()

```python
#Context mangager
with open("message.txt", mode ="r+") as file:
    print(file.read())
    file.write("This is first message \n")
```

```python
#Context mangager
with open("message.txt", mode ="w+") as file:
    print(file.read())
    file.write("In the begining \n")
```

### Filter()

```python
def even(num):
    return num% 2==0
```

```python
f_obj = filter(even, [1,3,4,6,8,9])
```

```python
team_tasks = [
    [True, True, True],       # Alice
    [True, False, True],      # Bob
    [False, True, True]       # Carol
]

names = ["Alice", "Bob", "Carol"]

for i in range(len(team_tasks)):
    if all(team_tasks[i]):
        print(f"{names[i]}: Task is done ")
    else:
        print(f"{names[i]}: Task is not done ")
```
