## Polymorphism in Python

### Operating Overloading

```python
class Point:
    """
    A class to represent a point in 2D space.
    Attributes: x: The x-coordinate of the point.
                y: The y-coordinate of the point.
    Methods: __init__(self,x,y): Initializes the point with x and y coordinates.
             __repr__(self): Returns a string representation of the point.
             __add__(self,other): Adds two points and returns a new point.
             __sub__(self,other): Subtracts two points and returns a new point.
             __pow__(self,p): Raises the point to the power of p and returns a new  point.
             __gt__(self,other): Compares two points and returns True if the current point is greater than the other point, False otherwise.
    """
    

    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point({self.x},{self.y})"

    def __add__(self,other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x,y)

    def __sub__(self,other):
        x = self.x - other.x
        y = self.y - other.y
        return Point(x,y)

    def __pow__(self,p):
        x = self.x ** p
        y = self.y ** p
        return Point(x,y)

    def __gt__(self,other):
        return self.x > other.x and self.y > other.y
    
```

```python
A=Point(2,4)
B=Point(5,6)
```

```python
A + B
```

```python
A - B
```

```python
A ** 3
```

```python
B>A
```

```python
print(A.__doc__)
```

### Duck Typing

```python
class Car:
    def drive(self):
        print("Driving a Car")

class Bike:
    def drive(self):
        print("Driving a bike")
    
```

```python
def start_driving(vehicle):
    vehicle.drive()
```

```python
car = Car()
bike = Bike()
```

```python
start_driving(bike)
```

```python
start_driving(car)
```
