### Class and Objects

```python
class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def sum(self):
        return self.num1 + self.num2

    def sub(self):
        return self.num1 - self.num2
        
```

```python
cal_obj = Calculator(10,4)
```

```python
type(cal_obj)
```

```python
cal_obj.sum()
```

```python
cal_obj2 = Calculator(55,7)
cal_obj2.sum()
```

```python
class Person:
    count = 0 #<Class attribute

    def __init__(self, name, address = "Kathmandu"): #<Constructer / Initializer
        """
        Creating a Person Class
        """
        self.name = name 
        self.address = address

        Person.count +=1

    def get_details(self): #< Instance Method
        print(f"I am {self.name} from {self.address}")

    @classmethod
    def get_count(cls):#< class method
        return cls.count

    @staticmethod
    def create_fullname(firstname, lastname): #< static method
        return f"{lastname}, {firstname}"
```

```python
student1 = Person("aayam")
```

```python
student2 = Person("Resason", "jhapa")
```

```python
student2.get_details()
```

```python
student2.get_count()
```

```python
student2.create_fullname("aayan","singh")
```

```python
class Library:
    def __init__(self, name:str, depart="Maths"):
        self.name = name
        self.depart = depart
        self.books = [
            ("The Alchemist", 25),
            ("The Da Vinci Code", 30),
            ("A Brief History of Time", 15),
            ("Angels & Demons", 0),
            ("The Grand Design", 0),
            ("1984", 19)
        ]
    
    def can_burrow(self, book_name):
        return any(quantity > 0 for book, quantity in self.books if book.lower() == book_name.lower())
book_obj1 = Library("The Alchemist")
book_obj1.can_burrow("The Alchemist")
 
```

```python
import statistics
class StudentResult:
    def __init__(self,name:str, marks:list):
        self.name = name 
        self.marks = marks

    def total(self):
        return sum(self.marks)

    def average(self):
        return statistics.mean(self.marks)
        

    def display(self):
        print(f"name is {self.name}")
        print(f"total is {self.total()}")
        print(f"average is {self.average()}")
 
stobj = StudentResult("aayam",[45,67,57,78])
stobj.total()
stobj.average()
stobj.display()
```
