## Method Overloading in Python

```python
class Example:
    def add(self, a, b=0, c=0):
        return a+b+c
```

```python
class Example:
    def add(self, *args):
        if len(args) == 1:
            return args[0]
        elif len(args) == 2:
            return args[0] + args[1]
        else:
            return "Invalid"

```

```python
ex = Example()
print(ex.add(1,3))
```
