## Abstraction in Python

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    def volume(self):
        return "not applicable for given shape"
        
```

```python
shape = Shape()
```

```python
class Rectangle(Shape):
    def __init__(self, width, length):
        self.w= width
        self.l= length

    def area(self):
        return self.w * self.l
        
```

```python
rect = Rectangle(32,34)
```

```python
rect.volume()
```

```python
rect.area()
```

```python
class Cuboid(Shape):
    def __init__(self, width, length, height):
        self.w = width
        self.l = length
        self.h = height

    def area(self):
        # Surface Area of Cuboid
        return 2 * (self.w * self.l + self.w * self.h + self.l * self.h)

    def volume(self):
        return self.w * self.l * self.h
```

```python
cuboid = Cuboid(2,3,4)
```

```python
cuboid.area()
```

```python
cuboid.volume()
```
