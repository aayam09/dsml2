## Method overriding in Python

```python
class Animal :
    def sound(self):
        print("Making Animal sound")
```

```python
class Dog(Animal):
    def sound(self):
        print(f"Dog makes woof sound")
```

```python
dog = Dog()
dog.sound()
```

```python
class Cat(Animal):
    def sound(self):
        print(f"Cat makes MEOW sound")
```

```python
cat = Cat()
cat.sound()
```

```python
class Parent:
    def say_hello(self):
        print("hello from Parent")

class Child(Parent):
    def say_hello(self):
        res = super().say_hello()
        print(res)
        print("Hello form child")
```

```python
obj = Child()
obj.say_hello()
```
