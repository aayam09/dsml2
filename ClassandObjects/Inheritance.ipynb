## inheritance in python

### 1.Single Inheritance

```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f"My name is {self.name}")
```

```python
class Cat(Animal):
    def __init__(self, name, age, color):
        super().__init__(name, age)
        self.color = color

    def speak(self):
        print("meow")
```

```python
cat = Cat("floppy", 3, "black")
```

```python
cat.name
```

```python
cat.speak()
```

```python
cat.intro()
```

### 2. Multiple Inheritance

```python
class Parent1:
    def func1(self):
        print("this is parent1")

class Parent2:
    def func2(self):
        print("this is parent2")

class Parent3:
    def func2(self):
        print("this is parent3")
```

```python
class Child(Parent1, Parent2, Parent3):
    def func4(self):
        print("This is child funtion")
```

```python
obj = Child()
```

```python
obj.func1()
```

```python
obj.func4()
```

```python
obj.func2()
```

### 3. Multi-level inheritance

```python
class A:
    def method_a(self):
        print("This is method from class A")

class B(A):
    def method_b(self):
        print("This is method from class B")

class C(B):
    def method_c(self):
        print("This is method from class C")
```

```python
obj = C()
```

```python
obj.method_a()
```

```python
obj.method_b()
```

```python
obj.method_c()
```

```python
class Person:
    def __init__(self, name, age):
        self.name, self.age = name, age

class Employee(Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id

class Manager(Employee):
    def __init__(self, name, age, employee_id, department):
        super().__init__(name, age, employee_id)
        self.department = department

    def getDetails(self):
        print(f"name: {self.name}\nage: {self.age}\nID: {self.employee_id}\ndepartment: {self.department}")
 
obj = Manager("aayam", 43, "332A", "accounts")
obj.getDetails()
 
```

### 4. Hierarchical Inheritance

```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def intro(self):
        print(f"My name is {self.name}")
```

```python
class Cat(Animal):
    def speak(self):
        return "meow"

class Lion(Animal):
    def speak(self):
        return "roar"

class Dog(Animal):
    def speak(self):
        return "woof"
```

```python
cat = Cat("Fluffy",13)
cat.speak()
```

```python
cat.intro()
```

### 5. Hybrid Inheritance

```python
class Animal:
    def __init__(self, name, age =2):
        self.name = name
        self.age = age

    def intro(self):
        print(f"My name is {self.name}")
```

```python
class Mammal(Animal):
    def __init__(self, name , fur = False):
        super().__init__(name)
        self.fur = fur

    def birth(self):
        print(f"{self.name} give birth to live ones")

class Birds(Animal):
    def fly(self):
        print(f"{self.name} can fly")
```

```python
class Bat(Mammal, Birds):
    pass
```

```python
b = Bat("Bruce")
```

```python
b.intro()
```

```python
b.birth()
```

```python
b.fly()
```
