```python
import time

class TimeSimulator:
    def countdown(self, seconds):
        try:
            if seconds <= 0:
                raise ValueError
            for x in range(seconds, 0, -1):
                time.sleep(1)
                print(f"{x} seconds remaining")
            print("Time's up!")
        except ValueError:
            print("Please provide a valid positive integer.")

simulator = TimeSimulator()
simulator.countdown(14)
 
```
