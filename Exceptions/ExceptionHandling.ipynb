## Basic Exception Handling in Python

```python
try:
    num = int(input("enter a number"))
    print(num ** 5)
except ValueError:
    print("Invalid Input, enter integer value")
    print(0)
```

## Catching Multiple Exception

```python
try:
    x = int(input("enter a number"))
    y = int(input("enter another number"))
    result = x/y

except ValueError as error:
    print("invalid input", error)
    result = 1

except ZeroDivisionError:
    print("Cannot divide by zero")
    result = 0

print("Final result:", result)
    
```

```python
lst = ['a', 'b', 'c', 'd']

def get_list_item(index):
    try:
        return lst[index]
    except IndexError:
        return lst[-1]
    except Exception:
        return lst

# Test cases to trigger all scenarios
print(get_list_item(2))          
print(get_list_item(5))          
print(get_list_item(-10))        
print(get_list_item('invalid'))  
 
```

## The Else Block

```python
try:
    x = int(input("enter a number"))
    y = int(input("enter another number"))
    result = x/y

except ValueError as error:
    print("invalid input", error)
    result = 1

except ZeroDivisionError:
    print("Cannot divide by zero")
    result = 0
else:
    print("Else block")
    print("Final result:", result)
    
```

## The finally block

```python
try:
    x = int(input("enter a number"))
    y = int(input("enter another number"))
    result = x/y

except ValueError as error:
    print("invalid input", error)
    result = 1

except ZeroDivisionError:
    print("Cannot divide by zero")
    result = 0

else:
    print("Else block")
    print("Final result:", result)

finally:
    print("final block")
    print("Final result:", result)

    
```
