## Python Generators

#### step1 : Define a generator function using yield keyword

```python
def number_generator(start, end):
    for num in range(start, end+1):
        yield num
```

#### step 2: Create a Generator object using the generator function

```python
gen = number_generator(10,23)
```

```python
gen
```

#### step 3: Use the genrator object to genrate the values on the fly

```python
next(gen)
```

#### step 4: Use a for loop to generate all the values in the sequence

```python
my_gen= number_generator(1,6)

for num in my_gen:
    print(num)
```

```python
## Given number is prime or not
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False

    return True


def prime_generator():
    n = 1
    while True:
        if is_prime(n):
            yield n
        n += 1
```

```python
prime_gen = prime_generator()
```

```python
for prime in prime_gen:
    print(prime)
    if prime > 100:
        break
```
