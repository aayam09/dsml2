## 2. Matrix operation

```python
a = np.array([[4,5,6],[1,2,3]])
b = np.random.randint(1,10,(3,3))
```

```python
a,b
```

```python
np.dot(a,b)
```

```python
np.dot(b,a.T)
```

```python
#a.T
np.transpose(a)
```

```python
np.trace(b)
```

```python
#Inverse of array
```

```python
np.linalg.det(b)
```

```python
np.linalg.inv(b)
```

```python
np.dot(
    b, np.linalg.inv(b)
).round()
```
