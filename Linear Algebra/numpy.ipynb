## Accesing an Array

```python
a = np.array([1,2,3,4,5])
```

```python
a[3]
```

```python
a[0:4:2]
```

```python
#2D
b = np.random.randint(0,25, size = (3,3))
```

```python
b
```

```python
b[1,0]
```

```python
b[-1,0]
```

```python
#slicing 2D array
b[:3,1]
```

```python
b[:,2]
```

```python
#3D
c = np.random.randint(0,25, size=(3,3,2))
```

```python
c
```

```python
c[0,2,1]
```

```python
c[:,2,:]
```

```python
a
```

```python
mask = a > 5
```

```python
mask
```

```python
a[mask]
```
