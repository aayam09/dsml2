## Why numpy over Python loop?

```python
import numpy as np
import time
```

```python
a = np.random.rand(90_000_000)
b = np.random.rand(90_000_000)
```

```python
st = time.time()
res = []
for _a, _b in zip(list(a),list(b)):
    res.append(_a*_b)
end = time.time()
```

```python
end - st
```

```python
# List Comprehension
#Using Pure Pyhton
st = time.time()
res = [_a*_b for _a, _b in zip(list(a),list(b))]
end = time.time()
```

```python
end - st
```

```python
# Map Object
st = time.time()
gen = map(lambda x,y: x*y, list(a), list(b))
res = list(gen)
end = time.time()
```

```python
end - st
```

```python
# Numpy operation 
st = time.time()
res = a*b
end = time.time()
```

```python
end - st
```
