### Set Operations

```python
A = {1,2,3,4}
B = {3,4,5,6}
```

```python
aub = A.union(B)
aub
```

```python
aNb = A.intersection(B)
aNb
```

```python
A_B = A.difference(B)
A_B
```

```python
B_A = B.difference(A)
B_A
```

```python
A.isdisjoint(B)
```

```python
A >= B
```
