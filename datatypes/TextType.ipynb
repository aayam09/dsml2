## string formatting

```python
name = "john"
age = 45
```

```python
print("your name is %s and you are %d years old" %(name.capitalize(), age))
```

```python
print("your name is {} and you are {} years old".format(name.capitalize(), age))
```

```python
print(f"your name is {name} and you are {age} years old")
```
