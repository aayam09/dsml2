## Python Comprehension

### 1. List Comprehension

```python
#usual way
squared = []
for num in range(1,10):
    squared.append(num ** 2)

print(squared)
```

```python
# list comprehension
squared = [num ** 2 for num in range(1,10)]
print(squared)
```

```python
# list comprehension
squared = [num for num in range(1,100) if num % 2 == 0]
print(squared)
```

```python
squared = [num**2 for num in range(1,50) if num % 2 != 0]
print(squared)
```

```python
adj = ['red','big','tasty']
fruits = ["apple", "banana", "cherry"]

pair =[(a,f) for a in adj for f in fruits]
print(pair)
```

```python
sent = "A single training set that has already been processed is usually split into several types of datasets in machine learning which is needed to check how well the training of the model went"

words = sent.split()

list =[]

for word in words:
    list.append(len(word))

print(list)
 
```

```python
count = [len(word) for word in words]
print(count)
```

### 2 Dictionary Comprehension

```python
#usual way
sqe_dict = {}

for num in range(1,5):
    sqe_dict[num] = num ** 2

print(sqe_dict)
```

```python
#Dict Comprehension
sqr_dict ={ num:num**2 for num in range(1,5)}
print(sqr_dict)
```

```python
#Dict Comprehension
sqr_dict ={ num:num**2 for num in range(1,10) if num%2==0}
print(sqr_dict)
```

```python
sent = "A single training set that has already been processed is usually split into several types of datasets in machine learning which is needed to check how well the training of the model went"


letterdict = {word : len(word) for word in sent.split() if len(word)>5}

print(letterdict)
```

```python
#Write a Python program that uses set comprehension to create a set of all unique vowels found in the following string: "Set comprehensions can be nested"

string = "Set comprehensions can be nested"
vowel_set = {letter for letter in string if letter.lower()  in set('aeiou')}
print(vowel_set)
```
