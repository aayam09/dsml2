## Python While loop

```python
i = 0

while i < 5:
    print(i)
    i += 1
```

#### Break statement

```python
i = 0

while i < 10:
    print(i)
    i += 1

    if i == 5:
        break
```

```python
for i in range(10):
    print(i)
    if i == 5:
        break
```

#### Continue statement

```python
names = ["saimon", "jesus", "john"]

for name in names:
    if name == "jesus":
        continue
    print(name)
```

```python
number= int(input("Enter a number"))
total_prod = 1
i =1
while i <=number:
    total_prod *= i
    i+=1
    
print(f"The sum of the first {number} natural numbers is: {total_prod}")
```
