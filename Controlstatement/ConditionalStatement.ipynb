## Python Conditional satatement

```python
x = 55

if x>0:
    print("inside of the block")
    print(f"{x} is greater than 0")

print("outside if block")
```

```python
x = 5
y = -5

if x>0 and y>0:
    print("both are positive")
```

```python
x =33 

if x%2 == 0:
    print("even")
else:
    print("odd")
```

```python
x = "madam"
if x == x[::-1]:
    print("palindrome")
```

```python
x = 6

if x < 0:
    print("X is negative")
elif(x == 0):
    print("x is zero")
else:
    print("x is positive")
```

```python
#WAP to grade a number (100-80) Grade A, (80-60) Grade B , (60-40) Grade C and less than 40 Grade D. [if and elif control statement]

number = 77
if number >= 80:
    print("grade A")
elif number >=60:
    print("grade B")
elif number >=40:
    print("grade C")
else:
    print("grade D")
```
