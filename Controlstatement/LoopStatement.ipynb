```python
adj = ['red','big','tasty']
fruits = ["apple", "banana", "cherry"]
```

```python
for a in adj:
    for f in fruits:
        print(a,f)
```

```python
person = {
    "name" : "vince",
    "age" : 30,
    "gender" : "male"
}
person_list = list(person.items())
person_list
```

```python
for each in person_list:
    print(each)
```

```python
for key,value in person_list:
    print(f"{key} = {value}")
```

```python
coord =[[1,2,3],[4,5,6],[7,8,9]]
print(coord)
```

```python
n = int(input("enter a number"))
for i in range(1,11):
    print(f"{i} * {n} = {i*n}")
```

```python
numbers = [2,5,3,6,7,33,5,8,68,0,12,4,6,66,9,101,55,32,89]

even_list, odd_list =[],[]

for x in numbers:
    if x % 2 == 0:
        even_list.append(x)
    else:
        odd_list.append(x)

print(f"evenlist is {even_list}")
print(f"oddlist is {odd_list}")
```
