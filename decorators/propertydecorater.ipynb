## Python Property Decorater

### Getter

```python
class Person:
    def __init__(self,name):
        self.__fullname = name
        
    @property
    def name(self):
        print("Getting Name")
        return self.__fullname
```

```python
person = Person("aayam")
```

```python
person.name
```

### Setter

```python
class Person:
    def __init__(self,name):
        self.__fullname = name
        
    @property
    def name(self):
        print("Getting Name")
        return self.__fullname

    @name.setter
    def name(self, value):
        print("Setting Name")
        self.__fullname = value
```

```python
person = Person("aayam")
```

```python
person.name
```

```python
person.name = "aayush"
```

```python
person.name
```

### Deleter

```python
class Person:
    def __init__(self,name):
        self.__fullname = name
        
    @property
    def name(self):
        print("Getting Name")
        return self.__fullname

    @name.setter
    def name(self, value):
        print("Setting Name")
        self.__fullname = value

    @name.deleter
    def name(self):
        print("Deleting Name")
        del self.__fullname
```

```python
person = Person("Aakash")
```

```python
person.name
```

```python
del person.name
```

```python
person.name
```

```python
#Write a decorator @timeit , that gives total time taken by function to run
 
import time

def timeit(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print(end - start)
    return wrapper

@timeit
def takes_time():
   time.sleep(4)
takes_time()
 
 
 
```
