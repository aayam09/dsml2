## Python Decoraters

```python
class Test:
    @staticmethod
    def something():
        print(1)
```

```python
# Simple Decorater function
def decorator(func):#< func as arguement
    #Inner function
    def wrapper():
        print("before Function call")
        func()#<Calling actual function
        print("after function call")

    return wrapper#< returning function as value
```

```python
@decorator
def greet():
    print("Hello, World")
```

```python
greet()
```

### Check email using decorater

```python
def validate_email(func):
    def wrapper(email,password):
        if email.endswith("@skill.com"):
            func(email, password)
        else:
            print("Only suppoted for skill")
    return wrapper
```

```python
@validate_email
def login_user(email, password):
    # check password validity
    print("Logged in")
```
